```python
import pandas as pd
from sklearn.model_selection import train_test_split
import os

df = pd.read_csv('../data/processed/hcm_housing_clean.csv')

# --- CHIA DỮ LIỆU LẦN 1 ---
# Lấy 70% cho tập Train, 30% còn lại gán tạm vào biến temp (để chia tiếp cho Dev/Test)
df_train, df_temp = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=df['district']
)

# --- CHIA DỮ LIỆU LẦN 2 ---
# Lấy 30% của tập temp chia đôi (test_size=0.5) -> Mỗi tập Dev và Test sẽ chiếm 15% tổng dữ liệu ban đầu
df_dev, df_test = train_test_split(
    df_temp,
    test_size=0.50,
    random_state=42,
    stratify=df_temp['district']
)

# 2. In thống kê để đối chiếu với "sheet Đọc trước"
print("--- THỐNG KÊ SỐ LƯỢNG MẪU ---")
print(f"Tổng số mẫu ban đầu: {len(df)}")
print(f"Tập Train (~70%): {len(df_train)} dòng")
print(f"Tập Dev   (~15%): {len(df_dev)} dòng")
print(f"Tập Test  (~15%): {len(df_test)} dòng")

# Đảm bảo thư mục lưu trữ tồn tại
os.makedirs('../data/processed', exist_ok=True)

# 3. Lưu ra các file CSV tương ứng
df_train.to_csv('../data/processed/train.csv', index=False)
df_dev.to_csv('../data/processed/dev.csv', index=False)
df_test.to_csv('../data/processed/test.csv', index=False)

print("\n[THÀNH CÔNG] Đã lưu 3 tập dữ liệu vào thư mục data/processed/")
print("🚨 CẢNH BÁO: TUYỆT ĐỐI KHÔNG AI ĐƯỢC MỞ HOẶC DÙNG FILE 'test.csv' CHO ĐẾN TASK T40! 🚨")
```
